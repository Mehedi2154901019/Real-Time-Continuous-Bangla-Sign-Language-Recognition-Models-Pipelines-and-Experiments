```python
# ============================================================
# BDSL47 - Signer-Independent Classification
# Landmark Normalization + Scaling Experiment
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score
)

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, BatchNormalization, LeakyReLU
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping


# ============================================================
# 1. SETTINGS
# ============================================================

DATA_DIR = (
    r"D:\aaa EAAI Major Revision\static one handed"
    r"\bdsl47_signer_independent"
    r"\final_bdsl47_dataset_csv"
)

TRAIN_FILE = os.path.join(DATA_DIR, "train.csv")
VAL_FILE   = os.path.join(DATA_DIR, "validation.csv")
TEST_FILE  = os.path.join(DATA_DIR, "test.csv")

RANDOM_SEED = 42

np.random.seed(RANDOM_SEED)
tf.random.set_seed(RANDOM_SEED)


# ============================================================
# NORMALIZATION SETTINGS
# ============================================================

APPLY_TRANSLATION_NORMALIZATION = True
APPLY_SCALE_NORMALIZATION = True
APPLY_ROTATION_NORMALIZATION = True


# ============================================================
# 2. LOAD DATA
# ============================================================

train_df = pd.read_csv(TRAIN_FILE)
val_df   = pd.read_csv(VAL_FILE)
test_df  = pd.read_csv(TEST_FILE)

print("=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

print(f"Train samples:      {len(train_df)}")
print(f"Validation samples: {len(val_df)}")
print(f"Test samples:       {len(test_df)}")

print(f"\nTrain shape:      {train_df.shape}")
print(f"Validation shape: {val_df.shape}")
print(f"Test shape:       {test_df.shape}")


# ============================================================
# 3. FEATURES AND LABEL
# ============================================================

FEATURE_COLUMNS = [
    f"{axis}{i:02d}"
    for i in range(21)
    for axis in ["x", "y", "z"]
]

LABEL_COLUMN = "Label"


# ============================================================
# 4. CHECK COLUMNS
# ============================================================

for name, df in [
    ("train", train_df),
    ("validation", val_df),
    ("test", test_df)
]:

    missing = [
        c for c in FEATURE_COLUMNS + [LABEL_COLUMN]
        if c not in df.columns
    ]

    if missing:
        raise ValueError(
            f"Missing columns in {name}.csv: {missing}"
        )


# ============================================================
# 5. EXTRACT X AND y
# ============================================================

X_train = train_df[FEATURE_COLUMNS].values.astype(np.float32)
y_train = train_df[LABEL_COLUMN].values.astype(np.int64)

X_val = val_df[FEATURE_COLUMNS].values.astype(np.float32)
y_val = val_df[LABEL_COLUMN].values.astype(np.int64)

X_test = test_df[FEATURE_COLUMNS].values.astype(np.float32)
y_test = test_df[LABEL_COLUMN].values.astype(np.int64)


# ============================================================
# 6. BASIC VALIDATION
# ============================================================

print("\n" + "=" * 70)
print("DATA VALIDATION")
print("=" * 70)

print(f"Number of features: {X_train.shape[1]}")

if X_train.shape[1] != 63:
    raise ValueError(
        f"Expected 63 features, found {X_train.shape[1]}"
    )

print(f"Number of classes: {len(np.unique(y_train))}")

print(f"Train labels:      {sorted(np.unique(y_train))}")
print(f"Validation labels: {sorted(np.unique(y_val))}")
print(f"Test labels:       {sorted(np.unique(y_test))}")

if np.isnan(X_train).any():
    raise ValueError("NaN values found in training data.")

if np.isnan(X_val).any():
    raise ValueError("NaN values found in validation data.")

if np.isnan(X_test).any():
    raise ValueError("NaN values found in test data.")


# ============================================================
# 7. RESHAPE
# ============================================================
#
# Current shape:
#
# (samples, 63)
#
# Convert to:
#
# (samples, 21, 3)
#
# Each landmark:
# [x, y, z]
#
# ============================================================

X_train = X_train.reshape(-1, 21, 3)
X_val   = X_val.reshape(-1, 21, 3)
X_test  = X_test.reshape(-1, 21, 3)


# ============================================================
# 8. LANDMARK NORMALIZATION FUNCTIONS
# ============================================================

def normalize_landmarks(X):
    """
    Apply per-sample:
        1. Translation normalization
        2. Scale normalization
        3. Rotation normalization

    Input:
        X -> (samples, 21, 3)

    Output:
        X -> (samples, 21, 3)
    """

    X = X.copy()

    for sample_idx in range(len(X)):

        landmarks = X[sample_idx]

        # ----------------------------------------------------
        # LANDMARK INDICES
        # ----------------------------------------------------
        #
        # 0  = Wrist
        # 5  = Index MCP
        # 9  = Middle MCP
        # 17 = Pinky MCP
        #
        # ----------------------------------------------------

        wrist = landmarks[0].copy()

        index_mcp = landmarks[5].copy()
        middle_mcp = landmarks[9].copy()
        pinky_mcp = landmarks[17].copy()


        # ====================================================
        # 1. TRANSLATION NORMALIZATION
        # ====================================================
        #
        # Move wrist to origin.
        #
        # New wrist = (0, 0, 0)
        #
        # ====================================================

        if APPLY_TRANSLATION_NORMALIZATION:

            landmarks = landmarks - wrist

            index_mcp = landmarks[5]
            middle_mcp = landmarks[9]
            pinky_mcp = landmarks[17]


        # ====================================================
        # 2. SCALE NORMALIZATION
        # ====================================================
        #
        # Use maximum distance from wrist as hand scale.
        #
        # This makes different hand sizes comparable.
        #
        # ====================================================

        if APPLY_SCALE_NORMALIZATION:

            distances = np.linalg.norm(
                landmarks,
                axis=1
            )

            scale = np.max(distances)

            if scale > 1e-8:
                landmarks = landmarks / scale


        # ====================================================
        # 3. ROTATION NORMALIZATION
        # ====================================================
        #
        # Construct a canonical hand coordinate system.
        #
        # u = wrist -> middle MCP
        #
        # v = index MCP direction orthogonalized
        #
        # w = cross product
        #
        # This reduces differences caused by hand orientation.
        #
        # ====================================================

        if APPLY_ROTATION_NORMALIZATION:

            wrist = landmarks[0]
            index_mcp = landmarks[5]
            middle_mcp = landmarks[9]
            pinky_mcp = landmarks[17]

            # ------------------------------------------------
            # First axis: wrist -> middle MCP
            # ------------------------------------------------

            axis_u = middle_mcp - wrist

            norm_u = np.linalg.norm(axis_u)

            if norm_u > 1e-8:

                axis_u = axis_u / norm_u

                # --------------------------------------------
                # Second axis:
                # index MCP direction
                # orthogonalized against axis_u
                # --------------------------------------------

                axis_v = index_mcp - wrist

                axis_v = (
                    axis_v
                    - np.dot(axis_v, axis_u) * axis_u
                )

                norm_v = np.linalg.norm(axis_v)

                if norm_v > 1e-8:

                    axis_v = axis_v / norm_v

                    # ----------------------------------------
                    # Third axis
                    # ----------------------------------------

                    axis_w = np.cross(
                        axis_u,
                        axis_v
                    )

                    norm_w = np.linalg.norm(axis_w)

                    if norm_w > 1e-8:

                        axis_w = axis_w / norm_w

                        # ------------------------------------
                        # Re-orthogonalize v
                        # ------------------------------------

                        axis_v = np.cross(
                            axis_w,
                            axis_u
                        )

                        axis_v = (
                            axis_v /
                            np.linalg.norm(axis_v)
                        )

                        # ------------------------------------
                        # Projection into canonical frame
                        # ------------------------------------

                        rotation_matrix = np.column_stack([
                            axis_v,
                            axis_u,
                            axis_w
                        ])

                        landmarks = (
                            landmarks @ rotation_matrix
                        )

        X[sample_idx] = landmarks

    return X


# ============================================================
# 9. APPLY GEOMETRIC NORMALIZATION
# ============================================================

print("\n" + "=" * 70)
print("LANDMARK NORMALIZATION")
print("=" * 70)

print(
    f"Translation normalization : "
    f"{APPLY_TRANSLATION_NORMALIZATION}"
)

print(
    f"Scale normalization       : "
    f"{APPLY_SCALE_NORMALIZATION}"
)

print(
    f"Rotation normalization    : "
    f"{APPLY_ROTATION_NORMALIZATION}"
)

print("\nApplying normalization...")


X_train = normalize_landmarks(X_train)
X_val   = normalize_landmarks(X_val)
X_test  = normalize_landmarks(X_test)

print("Normalization completed.")


# ============================================================
# 10. FLATTEN BACK TO 63 FEATURES
# ============================================================

X_train = X_train.reshape(-1, 63)
X_val   = X_val.reshape(-1, 63)
X_test  = X_test.reshape(-1, 63)


# ============================================================
# 11. STANDARDIZATION
# ============================================================
#
# IMPORTANT:
#
# StandardScaler is fitted ONLY on TRAIN.
#
# Validation and test are transformed using
# training statistics.
#
# ============================================================

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)

X_val = scaler.transform(X_val)

X_test = scaler.transform(X_test)

print("\n" + "=" * 70)
print("STANDARDIZATION")
print("=" * 70)

print("StandardScaler fitted on TRAINING data only.")
print("Validation and test transformed using training statistics.")


# ============================================================
# 12. MODEL
# ============================================================
#
# IMPORTANT:
# Same architecture as previous baseline.
#
# We change representation, NOT model capacity.
#
# ============================================================

def build_model():

    model = Sequential([

        Dense(
            128,
            input_shape=(63,)
        ),

        BatchNormalization(),

        LeakyReLU(),

        Dense(128),

        BatchNormalization(),

        LeakyReLU(),

        Dense(64),

        BatchNormalization(),

        LeakyReLU(),

        Dropout(0.33),

        Dense(32),

        BatchNormalization(),

        LeakyReLU(),

        Dropout(0.25),

        Dense(
            47,
            activation="softmax"
        )
    ])

    model.compile(
        optimizer=Adam(
            learning_rate=0.001
        ),

        loss="sparse_categorical_crossentropy",

        metrics=["accuracy"]
    )

    return model


model = build_model()


# ============================================================
# 13. MODEL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("MODEL ARCHITECTURE")
print("=" * 70)

model.summary()


# ============================================================
# 14. EARLY STOPPING
# ============================================================

early_stop = EarlyStopping(

    monitor="val_loss",

    patience=30,

    restore_best_weights=True,

    verbose=1
)


# ============================================================
# 15. TRAINING
# ============================================================

print("\n" + "=" * 70)
print("TRAINING")
print("=" * 70)

history = model.fit(

    X_train,

    y_train,

    validation_data=(
        X_val,
        y_val
    ),

    epochs=600,

    batch_size=64,

    callbacks=[
        early_stop
    ],

    verbose=1
)


# ============================================================
# 16. BEST EPOCH
# ============================================================

best_epoch = (
    np.argmin(
        history.history["val_loss"]
    ) + 1
)

print("\n" + "=" * 70)
print("TRAINING RESULT")
print("=" * 70)

print(
    f"Best epoch based on validation loss: "
    f"{best_epoch}"
)

print(
    f"Total epochs actually trained: "
    f"{len(history.history['loss'])}"
)


# ============================================================
# 17. LEARNING CURVES
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")

plt.title(
    "Training and Validation Accuracy"
)

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.show()


plt.figure(figsize=(10, 6))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.title(
    "Training and Validation Loss"
)

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.show()


# ============================================================
# 18. EVALUATION
# ============================================================

train_loss, train_acc = model.evaluate(
    X_train,
    y_train,
    verbose=0
)

val_loss, val_acc = model.evaluate(
    X_val,
    y_val,
    verbose=0
)

test_loss, test_acc = model.evaluate(
    X_test,
    y_test,
    verbose=0
)


# ============================================================
# 19. TEST PREDICTIONS
# ============================================================

y_test_prob = model.predict(
    X_test,
    verbose=0
)

y_test_pred = np.argmax(
    y_test_prob,
    axis=1
)


# ============================================================
# 20. METRICS
# ============================================================

macro_precision = precision_score(
    y_test,
    y_test_pred,
    average="macro",
    zero_division=0
)

macro_recall = recall_score(
    y_test,
    y_test_pred,
    average="macro",
    zero_division=0
)

macro_f1 = f1_score(
    y_test,
    y_test_pred,
    average="macro",
    zero_division=0
)

weighted_precision = precision_score(
    y_test,
    y_test_pred,
    average="weighted",
    zero_division=0
)

weighted_recall = recall_score(
    y_test,
    y_test_pred,
    average="weighted",
    zero_division=0
)

weighted_f1 = f1_score(
    y_test,
    y_test_pred,
    average="weighted",
    zero_division=0
)


# ============================================================
# 21. FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 70)
print("FINAL SIGNER-INDEPENDENT EXPERIMENT SUMMARY")
print("=" * 70)

print(f"Train samples:       {len(y_train)}")
print(f"Validation samples:  {len(y_val)}")
print(f"Test samples:        {len(y_test)}")

print(f"\nBest epoch:           {best_epoch}")

print("\nNormalization:")
print(
    f"Translation:         "
    f"{APPLY_TRANSLATION_NORMALIZATION}"
)

print(
    f"Scale:                "
    f"{APPLY_SCALE_NORMALIZATION}"
)

print(
    f"Rotation:             "
    f"{APPLY_ROTATION_NORMALIZATION}"
)

print("\nAccuracy:")

print(
    f"Train Accuracy:      "
    f"{train_acc * 100:.2f}%"
)

print(
    f"Validation Accuracy: "
    f"{val_acc * 100:.2f}%"
)

print(
    f"Test Accuracy:       "
    f"{test_acc * 100:.2f}%"
)

print("\nMacro Metrics:")

print(
    f"Macro Precision:     "
    f"{macro_precision:.4f}"
)

print(
    f"Macro Recall:        "
    f"{macro_recall:.4f}"
)

print(
    f"Macro F1-score:      "
    f"{macro_f1:.4f}"
)

print("\nWeighted Metrics:")

print(
    f"Weighted Precision:  "
    f"{weighted_precision:.4f}"
)

print(
    f"Weighted Recall:     "
    f"{weighted_recall:.4f}"
)

print(
    f"Weighted F1-score:   "
    f"{weighted_f1:.4f}"
)


# ============================================================
# 22. CLASSIFICATION REPORT
# ============================================================

print("\n" + "=" * 70)
print("FINAL TEST SET CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        y_test,
        y_test_pred,
        digits=4,
        zero_division=0
    )
)


# ============================================================
# 23. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    y_test_pred,
    labels=np.arange(47)
)

plt.figure(figsize=(12, 10))

plt.imshow(
    cm,
    interpolation="nearest"
)

plt.title(
    "Confusion Matrix - "
    "Signer-Independent Test Set"
)

plt.xlabel("Predicted Label")

plt.ylabel("True Label")

plt.colorbar()

plt.xticks(
    np.arange(47),
    np.arange(47),
    rotation=90
)

plt.yticks(
    np.arange(47),
    np.arange(47)
)

plt.tight_layout()

plt.show()


# ============================================================
# 24. NORMALIZED CONFUSION MATRIX
# ============================================================

cm_normalized = (
    cm.astype(float)
    / cm.sum(
        axis=1,
        keepdims=True
    )
)

cm_normalized = np.nan_to_num(
    cm_normalized
)

plt.figure(figsize=(12, 10))

plt.imshow(
    cm_normalized,
    interpolation="nearest"
)

plt.title(
    "Normalized Confusion Matrix - "
    "Signer-Independent Test Set"
)

plt.xlabel("Predicted Label")

plt.ylabel("True Label")

plt.colorbar()

plt.xticks(
    np.arange(47),
    np.arange(47),
    rotation=90
)

plt.yticks(
    np.arange(47),
    np.arange(47)
)

plt.tight_layout()

plt.show()


# ============================================================
# 25. PER-CLASS F1
# ============================================================

per_class_f1 = f1_score(
    y_test,
    y_test_pred,
    average=None,
    labels=np.arange(47),
    zero_division=0
)

print("\n" + "=" * 70)
print("PER-CLASS F1 SCORES")
print("=" * 70)

for class_id, score in enumerate(per_class_f1):

    print(
        f"Class {class_id:2d}: "
        f"F1 = {score:.4f}"
    )


# ============================================================
# 26. SAVE MODEL
# ============================================================

model_path = os.path.join(
    DATA_DIR,
    "bdsl47_signer_independent_normalized_mlp.keras"
)

model.save(model_path)

print("\n" + "=" * 70)
print("MODEL SAVED")
print("=" * 70)

print(model_path)

print("\nExperiment completed successfully.")
```

```python
# ============================================================
# BDSL47 - Signer-Independent Classification
# Translation + Scale Normalization
# + Training-Only Landmark Augmentation
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score
)

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, BatchNormalization, LeakyReLU
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping


# ============================================================
# 1. SETTINGS
# ============================================================

DATA_DIR = (
    r"D:\aaa EAAI Major Revision\static one handed"
    r"\bdsl47_signer_independent"
    r"\final_bdsl47_dataset_csv"
)

TRAIN_FILE = os.path.join(DATA_DIR, "train.csv")
VAL_FILE   = os.path.join(DATA_DIR, "validation.csv")
TEST_FILE  = os.path.join(DATA_DIR, "test.csv")

RANDOM_SEED = 42

np.random.seed(RANDOM_SEED)
tf.random.set_seed(RANDOM_SEED)


# ============================================================
# 2. AUGMENTATION SETTINGS
# ============================================================
#
# IMPORTANT:
# Augmentation is applied ONLY to training data.
#
# The original training samples are retained.
#
# AUGMENTATION_FACTOR = 1 means:
#   original + 1 augmented copy
#
# Therefore training size approximately doubles.
#
# ============================================================

AUGMENTATION_FACTOR = 1

# Maximum random rotation in degrees
MAX_ROTATION_DEGREES = 10.0

# Random scale variation
SCALE_MIN = 0.95
SCALE_MAX = 1.05

# Landmark coordinate noise
NOISE_STD = 0.005


# ============================================================
# 3. LOAD DATA
# ============================================================

train_df = pd.read_csv(TRAIN_FILE)
val_df   = pd.read_csv(VAL_FILE)
test_df  = pd.read_csv(TEST_FILE)

print("=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

print(f"Train samples:      {len(train_df)}")
print(f"Validation samples: {len(val_df)}")
print(f"Test samples:       {len(test_df)}")


# ============================================================
# 4. FEATURES
# ============================================================

FEATURE_COLUMNS = [
    f"{axis}{i:02d}"
    for i in range(21)
    for axis in ["x", "y", "z"]
]

LABEL_COLUMN = "Label"


# ============================================================
# 5. CHECK COLUMNS
# ============================================================

for name, df in [
    ("train", train_df),
    ("validation", val_df),
    ("test", test_df)
]:

    missing = [
        c for c in FEATURE_COLUMNS + [LABEL_COLUMN]
        if c not in df.columns
    ]

    if missing:
        raise ValueError(
            f"Missing columns in {name}.csv: {missing}"
        )


# ============================================================
# 6. LOAD X AND y
# ============================================================

X_train = train_df[FEATURE_COLUMNS].values.astype(np.float32)
y_train = train_df[LABEL_COLUMN].values.astype(np.int64)

X_val = val_df[FEATURE_COLUMNS].values.astype(np.float32)
y_val = val_df[LABEL_COLUMN].values.astype(np.int64)

X_test = test_df[FEATURE_COLUMNS].values.astype(np.float32)
y_test = test_df[LABEL_COLUMN].values.astype(np.int64)


# ============================================================
# 7. BASIC VALIDATION
# ============================================================

print("\n" + "=" * 70)
print("DATA VALIDATION")
print("=" * 70)

print(f"Number of features: {X_train.shape[1]}")

if X_train.shape[1] != 63:
    raise ValueError(
        f"Expected 63 features, found {X_train.shape[1]}"
    )

print(f"Number of training classes: {len(np.unique(y_train))}")
print(f"Train labels:      {sorted(np.unique(y_train))}")
print(f"Validation labels: {sorted(np.unique(y_val))}")
print(f"Test labels:       {sorted(np.unique(y_test))}")

if np.isnan(X_train).any():
    raise ValueError("NaN values found in training data.")

if np.isnan(X_val).any():
    raise ValueError("NaN values found in validation data.")

if np.isnan(X_test).any():
    raise ValueError("NaN values found in test data.")


# ============================================================
# 8. RESHAPE TO 21 LANDMARKS × 3 COORDINATES
# ============================================================

X_train = X_train.reshape(-1, 21, 3)
X_val   = X_val.reshape(-1, 21, 3)
X_test  = X_test.reshape(-1, 21, 3)


# ============================================================
# 9. TRANSLATION + SCALE NORMALIZATION
# ============================================================

def normalize_landmarks(X):

    X = X.copy()

    for i in range(len(X)):

        landmarks = X[i]

        # ----------------------------------------------------
        # Translation normalization
        # Wrist (landmark 0) becomes origin
        # ----------------------------------------------------

        wrist = landmarks[0].copy()

        landmarks = landmarks - wrist

        # ----------------------------------------------------
        # Scale normalization
        #
        # Maximum distance from wrist is used as scale.
        # ----------------------------------------------------

        distances = np.linalg.norm(
            landmarks,
            axis=1
        )

        scale = np.max(distances)

        if scale > 1e-8:
            landmarks = landmarks / scale

        X[i] = landmarks

    return X


print("\n" + "=" * 70)
print("LANDMARK NORMALIZATION")
print("=" * 70)

print("Translation normalization: ENABLED")
print("Scale normalization:       ENABLED")
print("Rotation canonicalization: DISABLED")

X_train = normalize_landmarks(X_train)
X_val   = normalize_landmarks(X_val)
X_test  = normalize_landmarks(X_test)


# ============================================================
# 10. LANDMARK AUGMENTATION
# ============================================================
#
# Applied ONLY to normalized TRAINING landmarks.
#
# Each augmented sample receives:
#
#   1. Small random 3D rotation
#   2. Small random scale change
#   3. Small Gaussian coordinate noise
#
# The gesture label remains unchanged.
#
# ============================================================

def rotation_matrix_x(angle):
    c = np.cos(angle)
    s = np.sin(angle)

    return np.array([
        [1, 0, 0],
        [0, c, -s],
        [0, s, c]
    ], dtype=np.float32)


def rotation_matrix_y(angle):
    c = np.cos(angle)
    s = np.sin(angle)

    return np.array([
        [c, 0, s],
        [0, 1, 0],
        [-s, 0, c]
    ], dtype=np.float32)


def rotation_matrix_z(angle):
    c = np.cos(angle)
    s = np.sin(angle)

    return np.array([
        [c, -s, 0],
        [s, c, 0],
        [0, 0, 1]
    ], dtype=np.float32)


def augment_landmarks(X, y):

    augmented_X = []
    augmented_y = []

    max_angle = np.deg2rad(
        MAX_ROTATION_DEGREES
    )

    for sample, label in zip(X, y):

        # ----------------------------------------------------
        # Random rotation
        # ----------------------------------------------------

        angle_x = np.random.uniform(
            -max_angle,
            max_angle
        )

        angle_y = np.random.uniform(
            -max_angle,
            max_angle
        )

        angle_z = np.random.uniform(
            -max_angle,
            max_angle
        )

        Rx = rotation_matrix_x(angle_x)
        Ry = rotation_matrix_y(angle_y)
        Rz = rotation_matrix_z(angle_z)

        R = Rz @ Ry @ Rx

        augmented = sample @ R.T

        # ----------------------------------------------------
        # Random scale
        # ----------------------------------------------------

        random_scale = np.random.uniform(
            SCALE_MIN,
            SCALE_MAX
        )

        augmented = augmented * random_scale

        # ----------------------------------------------------
        # Small landmark noise
        # ----------------------------------------------------

        noise = np.random.normal(
            loc=0.0,
            scale=NOISE_STD,
            size=augmented.shape
        ).astype(np.float32)

        augmented = augmented + noise

        augmented_X.append(augmented)
        augmented_y.append(label)

    return (
        np.asarray(augmented_X, dtype=np.float32),
        np.asarray(augmented_y, dtype=np.int64)
    )


# ============================================================
# 11. CREATE AUGMENTED TRAINING DATA
# ============================================================

print("\n" + "=" * 70)
print("TRAINING AUGMENTATION")
print("=" * 70)

print(
    f"Augmentation factor: "
    f"{AUGMENTATION_FACTOR}"
)

print(
    f"Maximum rotation: "
    f"±{MAX_ROTATION_DEGREES} degrees"
)

print(
    f"Scale range: "
    f"{SCALE_MIN} - {SCALE_MAX}"
)

print(
    f"Noise standard deviation: "
    f"{NOISE_STD}"
)

augmented_X_parts = []
augmented_y_parts = []

for i in range(AUGMENTATION_FACTOR):

    X_aug, y_aug = augment_landmarks(
        X_train,
        y_train
    )

    augmented_X_parts.append(X_aug)
    augmented_y_parts.append(y_aug)

if AUGMENTATION_FACTOR > 0:

    X_augmented = np.concatenate(
        augmented_X_parts,
        axis=0
    )

    y_augmented = np.concatenate(
        augmented_y_parts,
        axis=0
    )

else:

    X_augmented = np.empty(
        (0, 21, 3),
        dtype=np.float32
    )

    y_augmented = np.empty(
        (0,),
        dtype=np.int64
    )


# ============================================================
# 12. COMBINE ORIGINAL + AUGMENTED TRAINING DATA
# ============================================================

X_train_combined = np.concatenate(
    [
        X_train,
        X_augmented
    ],
    axis=0
)

y_train_combined = np.concatenate(
    [
        y_train,
        y_augmented
    ],
    axis=0
)

print("\nOriginal training samples:")
print(f"{len(X_train):,}")

print("\nAugmented samples:")
print(f"{len(X_augmented):,}")

print("\nFinal training samples:")
print(f"{len(X_train_combined):,}")


# ============================================================
# 13. FLATTEN TO 63 FEATURES
# ============================================================

X_train_combined = X_train_combined.reshape(-1, 63)

X_val = X_val.reshape(-1, 63)

X_test = X_test.reshape(-1, 63)


# ============================================================
# 14. STANDARDIZATION
# ============================================================
#
# FIT ONLY ON THE ORIGINAL TRAINING DATA.
#
# This is deliberately done using X_train, NOT validation/test.
#
# Augmented training samples are then transformed using the
# same training-derived scaler.
#
# ============================================================

scaler = StandardScaler()

scaler.fit(
    X_train
    .reshape(-1, 63)
)

X_train_combined = scaler.transform(
    X_train_combined
)

X_val = scaler.transform(
    X_val
)

X_test = scaler.transform(
    X_test
)

print("\n" + "=" * 70)
print("STANDARDIZATION")
print("=" * 70)

print(
    "StandardScaler fitted on ORIGINAL TRAINING data only."
)

print(
    "Augmented training, validation and test data "
    "transformed using training statistics."
)


# ============================================================
# 15. MODEL
# ============================================================
#
# SAME ARCHITECTURE AS PREVIOUS EXPERIMENT.
#
# ============================================================

def build_model():

    model = Sequential([

        Dense(
            128,
            input_shape=(63,)
        ),

        BatchNormalization(),

        LeakyReLU(),

        Dense(128),

        BatchNormalization(),

        LeakyReLU(),

        Dense(64),

        BatchNormalization(),

        LeakyReLU(),

        Dropout(0.33),

        Dense(32),

        BatchNormalization(),

        LeakyReLU(),

        Dropout(0.25),

        Dense(
            47,
            activation="softmax"
        )
    ])

    model.compile(

        optimizer=Adam(
            learning_rate=0.001
        ),

        loss="sparse_categorical_crossentropy",

        metrics=["accuracy"]
    )

    return model


model = build_model()


# ============================================================
# 16. MODEL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("MODEL ARCHITECTURE")
print("=" * 70)

model.summary()


# ============================================================
# 17. EARLY STOPPING
# ============================================================

early_stop = EarlyStopping(

    monitor="val_loss",

    patience=30,

    restore_best_weights=True,

    verbose=1
)


# ============================================================
# 18. TRAINING
# ============================================================

print("\n" + "=" * 70)
print("TRAINING")
print("=" * 70)

history = model.fit(

    X_train_combined,

    y_train_combined,

    validation_data=(
        X_val,
        y_val
    ),

    epochs=600,

    batch_size=64,

    callbacks=[
        early_stop
    ],

    verbose=1
)


# ============================================================
# 19. BEST EPOCH
# ============================================================

best_epoch = (
    np.argmin(
        history.history["val_loss"]
    ) + 1
)

print("\n" + "=" * 70)
print("TRAINING RESULT")
print("=" * 70)

print(
    f"Best epoch based on validation loss: "
    f"{best_epoch}"
)

print(
    f"Total epochs actually trained: "
    f"{len(history.history['loss'])}"
)


# ============================================================
# 20. LEARNING CURVES
# ============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")

plt.title(
    "Training and Validation Accuracy"
)

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.show()


plt.figure(figsize=(10, 6))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.title(
    "Training and Validation Loss"
)

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.show()


# ============================================================
# 21. EVALUATION
# ============================================================

train_loss, train_acc = model.evaluate(
    X_train_combined,
    y_train_combined,
    verbose=0
)

val_loss, val_acc = model.evaluate(
    X_val,
    y_val,
    verbose=0
)

test_loss, test_acc = model.evaluate(
    X_test,
    y_test,
    verbose=0
)


# ============================================================
# 22. TEST PREDICTIONS
# ============================================================

y_test_prob = model.predict(
    X_test,
    verbose=0
)

y_test_pred = np.argmax(
    y_test_prob,
    axis=1
)


# ============================================================
# 23. METRICS
# ============================================================

macro_precision = precision_score(
    y_test,
    y_test_pred,
    average="macro",
    zero_division=0
)

macro_recall = recall_score(
    y_test,
    y_test_pred,
    average="macro",
    zero_division=0
)

macro_f1 = f1_score(
    y_test,
    y_test_pred,
    average="macro",
    zero_division=0
)

weighted_precision = precision_score(
    y_test,
    y_test_pred,
    average="weighted",
    zero_division=0
)

weighted_recall = recall_score(
    y_test,
    y_test_pred,
    average="weighted",
    zero_division=0
)

weighted_f1 = f1_score(
    y_test,
    y_test_pred,
    average="weighted",
    zero_division=0
)


# ============================================================
# 24. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("FINAL SIGNER-INDEPENDENT EXPERIMENT SUMMARY")
print("=" * 70)

print(f"Original train samples:  {len(y_train):,}")
print(f"Augmented train samples: {len(y_augmented):,}")
print(f"Final train samples:     {len(y_train_combined):,}")

print(f"Validation samples:      {len(y_val):,}")
print(f"Test samples:            {len(y_test):,}")

print(f"\nBest epoch: {best_epoch}")

print("\nPreprocessing:")
print("Translation normalization : YES")
print("Scale normalization       : YES")
print("Rotation canonicalization : NO")
print("Training augmentation     : YES")

print("\nAugmentation:")
print(
    f"Rotation: ±{MAX_ROTATION_DEGREES} degrees"
)
print(
    f"Scale: {SCALE_MIN}–{SCALE_MAX}"
)
print(
    f"Noise STD: {NOISE_STD}"
)

print("\nAccuracy:")

print(
    f"Train Accuracy:      "
    f"{train_acc * 100:.2f}%"
)

print(
    f"Validation Accuracy: "
    f"{val_acc * 100:.2f}%"
)

print(
    f"Test Accuracy:       "
    f"{test_acc * 100:.2f}%"
)

print("\nMacro Metrics:")

print(
    f"Macro Precision:     "
    f"{macro_precision:.4f}"
)

print(
    f"Macro Recall:        "
    f"{macro_recall:.4f}"
)

print(
    f"Macro F1-score:      "
    f"{macro_f1:.4f}"
)

print("\nWeighted Metrics:")

print(
    f"Weighted Precision:  "
    f"{weighted_precision:.4f}"
)

print(
    f"Weighted Recall:     "
    f"{weighted_recall:.4f}"
)

print(
    f"Weighted F1-score:   "
    f"{weighted_f1:.4f}"
)


# ============================================================
# 25. CLASSIFICATION REPORT
# ============================================================

print("\n" + "=" * 70)
print("FINAL TEST SET CLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        y_test,
        y_test_pred,
        digits=4,
        zero_division=0
    )
)


# ============================================================
# 26. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    y_test_pred,
    labels=np.arange(47)
)

plt.figure(figsize=(12, 10))

plt.imshow(
    cm,
    interpolation="nearest"
)

plt.title(
    "Confusion Matrix - "
    "Signer-Independent Test Set"
)

plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.colorbar()

plt.xticks(
    np.arange(47),
    np.arange(47),
    rotation=90
)

plt.yticks(
    np.arange(47),
    np.arange(47)
)

plt.tight_layout()

plt.show()


# ============================================================
# 27. NORMALIZED CONFUSION MATRIX
# ============================================================

cm_normalized = (
    cm.astype(float)
    / cm.sum(
        axis=1,
        keepdims=True
    )
)

cm_normalized = np.nan_to_num(
    cm_normalized
)

plt.figure(figsize=(12, 10))

plt.imshow(
    cm_normalized,
    interpolation="nearest"
)

plt.title(
    "Normalized Confusion Matrix - "
    "Signer-Independent Test Set"
)

plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.colorbar()

plt.xticks(
    np.arange(47),
    np.arange(47),
    rotation=90
)

plt.yticks(
    np.arange(47),
    np.arange(47)
)

plt.tight_layout()

plt.show()


# ============================================================
# 28. PER-CLASS F1
# ============================================================

per_class_f1 = f1_score(
    y_test,
    y_test_pred,
    average=None,
    labels=np.arange(47),
    zero_division=0
)

print("\n" + "=" * 70)
print("PER-CLASS F1 SCORES")
print("=" * 70)

for class_id, score in enumerate(per_class_f1):

    print(
        f"Class {class_id:2d}: "
        f"F1 = {score:.4f}"
    )


# ============================================================
# 29. SAVE MODEL
# ============================================================

model_path = os.path.join(
    DATA_DIR,
    "bdsl47_signer_independent_augmented_mlp.keras"
)

model.save(model_path)

print("\n" + "=" * 70)
print("MODEL SAVED")
print("=" * 70)

print(model_path)

print("\nExperiment completed successfully.")
```

```python
# ============================================================
# BDSL47 - Signer-Independent Classification
# Translation + Scale Normalization
# + Training-Only Landmark Augmentation
# Models: KNN & SVM
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score,
    accuracy_score
)

# ============================================================
# 1. SETTINGS
# ============================================================

DATA_DIR = (
    r"D:\aaa EAAI Major Revision\static one handed"
    r"\bdsl47_signer_independent"
    r"\final_bdsl47_dataset_csv"
)

TRAIN_FILE = os.path.join(DATA_DIR, "train.csv")
VAL_FILE   = os.path.join(DATA_DIR, "validation.csv")
TEST_FILE  = os.path.join(DATA_DIR, "test.csv")

RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)


# ============================================================
# 2. AUGMENTATION SETTINGS
# ============================================================

AUGMENTATION_FACTOR = 1
MAX_ROTATION_DEGREES = 10.0
SCALE_MIN = 0.95
SCALE_MAX = 1.05
NOISE_STD = 0.005


# ============================================================
# 3. LOAD DATA
# ============================================================

train_df = pd.read_csv(TRAIN_FILE)
val_df   = pd.read_csv(VAL_FILE)
test_df  = pd.read_csv(TEST_FILE)

print("=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

print(f"Train samples:      {len(train_df)}")
print(f"Validation samples: {len(val_df)}")
print(f"Test samples:       {len(test_df)}")


# ============================================================
# 4. FEATURES
# ============================================================

FEATURE_COLUMNS = [
    f"{axis}{i:02d}"
    for i in range(21)
    for axis in ["x", "y", "z"]
]

LABEL_COLUMN = "Label"


# ============================================================
# 5. CHECK COLUMNS
# ============================================================

for name, df in [
    ("train", train_df),
    ("validation", val_df),
    ("test", test_df)
]:
    missing = [
        c for c in FEATURE_COLUMNS + [LABEL_COLUMN]
        if c not in df.columns
    ]
    if missing:
        raise ValueError(
            f"Missing columns in {name}.csv: {missing}"
        )


# ============================================================
# 6. LOAD X AND y
# ============================================================

X_train = train_df[FEATURE_COLUMNS].values.astype(np.float32)
y_train = train_df[LABEL_COLUMN].values.astype(np.int64)

X_val = val_df[FEATURE_COLUMNS].values.astype(np.float32)
y_val = val_df[LABEL_COLUMN].values.astype(np.int64)

X_test = test_df[FEATURE_COLUMNS].values.astype(np.float32)
y_test = test_df[LABEL_COLUMN].values.astype(np.int64)


# ============================================================
# 7. BASIC VALIDATION
# ============================================================

print("\n" + "=" * 70)
print("DATA VALIDATION")
print("=" * 70)

print(f"Number of features: {X_train.shape[1]}")

if X_train.shape[1] != 63:
    raise ValueError(
        f"Expected 63 features, found {X_train.shape[1]}"
    )

print(f"Number of training classes: {len(np.unique(y_train))}")
print(f"Train labels:      {sorted(np.unique(y_train))}")
print(f"Validation labels: {sorted(np.unique(y_val))}")
print(f"Test labels:       {sorted(np.unique(y_test))}")

if np.isnan(X_train).any() or np.isnan(X_val).any() or np.isnan(X_test).any():
    raise ValueError("NaN values found in data.")


# ============================================================
# 8. RESHAPE TO 21 LANDMARKS × 3 COORDINATES
# ============================================================

X_train = X_train.reshape(-1, 21, 3)
X_val   = X_val.reshape(-1, 21, 3)
X_test  = X_test.reshape(-1, 21, 3)


# ============================================================
# 9. TRANSLATION + SCALE NORMALIZATION
# ============================================================

def normalize_landmarks(X):
    X = X.copy()
    for i in range(len(X)):
        landmarks = X[i]
        wrist = landmarks[0].copy()
        landmarks = landmarks - wrist
        distances = np.linalg.norm(landmarks, axis=1)
        scale = np.max(distances)
        if scale > 1e-8:
            landmarks = landmarks / scale
        X[i] = landmarks
    return X


print("\n" + "=" * 70)
print("LANDMARK NORMALIZATION")
print("=" * 70)

print("Translation normalization: ENABLED")
print("Scale normalization:       ENABLED")
print("Rotation canonicalization: DISABLED")

X_train = normalize_landmarks(X_train)
X_val   = normalize_landmarks(X_val)
X_test  = normalize_landmarks(X_test)


# ============================================================
# 10. LANDMARK AUGMENTATION
# ============================================================

def rotation_matrix_x(angle):
    c = np.cos(angle)
    s = np.sin(angle)
    return np.array([[1, 0, 0], [0, c, -s], [0, s, c]], dtype=np.float32)

def rotation_matrix_y(angle):
    c = np.cos(angle)
    s = np.sin(angle)
    return np.array([[c, 0, s], [0, 1, 0], [-s, 0, c]], dtype=np.float32)

def rotation_matrix_z(angle):
    c = np.cos(angle)
    s = np.sin(angle)
    return np.array([[c, -s, 0], [s, c, 0], [0, 0, 1]], dtype=np.float32)

def augment_landmarks(X, y):
    augmented_X = []
    augmented_y = []
    max_angle = np.deg2rad(MAX_ROTATION_DEGREES)

    for sample, label in zip(X, y):
        angle_x = np.random.uniform(-max_angle, max_angle)
        angle_y = np.random.uniform(-max_angle, max_angle)
        angle_z = np.random.uniform(-max_angle, max_angle)

        R = rotation_matrix_z(angle_z) @ rotation_matrix_y(angle_y) @ rotation_matrix_x(angle_x)
        augmented = sample @ R.T

        random_scale = np.random.uniform(SCALE_MIN, SCALE_MAX)
        augmented = augmented * random_scale

        noise = np.random.normal(loc=0.0, scale=NOISE_STD, size=augmented.shape).astype(np.float32)
        augmented = augmented + noise

        augmented_X.append(augmented)
        augmented_y.append(label)

    return np.asarray(augmented_X, dtype=np.float32), np.asarray(augmented_y, dtype=np.int64)


# ============================================================
# 11. CREATE AUGMENTED TRAINING DATA
# ============================================================

print("\n" + "=" * 70)
print("TRAINING AUGMENTATION")
print("=" * 70)

print(f"Augmentation factor: {AUGMENTATION_FACTOR}")
print(f"Maximum rotation:    ±{MAX_ROTATION_DEGREES} degrees")
print(f"Scale range:         {SCALE_MIN} - {SCALE_MAX}")
print(f"Noise std dev:       {NOISE_STD}")

augmented_X_parts = []
augmented_y_parts = []

for i in range(AUGMENTATION_FACTOR):
    X_aug, y_aug = augment_landmarks(X_train, y_train)
    augmented_X_parts.append(X_aug)
    augmented_y_parts.append(y_aug)

if AUGMENTATION_FACTOR > 0:
    X_augmented = np.concatenate(augmented_X_parts, axis=0)
    y_augmented = np.concatenate(augmented_y_parts, axis=0)
else:
    X_augmented = np.empty((0, 21, 3), dtype=np.float32)
    y_augmented = np.empty((0,), dtype=np.int64)


# ============================================================
# 12. COMBINE ORIGINAL + AUGMENTED TRAINING DATA
# ============================================================

X_train_combined = np.concatenate([X_train, X_augmented], axis=0)
y_train_combined = np.concatenate([y_train, y_augmented], axis=0)

print(f"\nOriginal training samples: {len(X_train):,}")
print(f"Augmented samples:         {len(X_augmented):,}")
print(f"Final training samples:    {len(X_train_combined):,}")


# ============================================================
# 13. FLATTEN TO 63 FEATURES
# ============================================================

X_train_combined = X_train_combined.reshape(-1, 63)
X_val = X_val.reshape(-1, 63)
X_test = X_test.reshape(-1, 63)


# ============================================================
# 14. STANDARDIZATION
# ============================================================

scaler = StandardScaler()
scaler.fit(X_train.reshape(-1, 63))

X_train_combined = scaler.transform(X_train_combined)
X_val = scaler.transform(X_val)
X_test = scaler.transform(X_test)

print("\n" + "=" * 70)
print("STANDARDIZATION")
print("=" * 70)
print("StandardScaler fitted on ORIGINAL TRAINING data only.")


# ============================================================
# 15. TRAIN AND EVALUATE: K-NEAREST NEIGHBORS (KNN)
# ============================================================

print("\n" + "=" * 70)
print("TRAINING KNN CLASSIFIER")
print("=" * 70)

knn_model = KNeighborsClassifier(n_neighbors=5, weights='distance', n_jobs=-1)
knn_model.fit(X_train_combined, y_train_combined)

knn_train_acc = accuracy_score(y_train_combined, knn_model.predict(X_train_combined))
knn_val_acc = accuracy_score(y_val, knn_model.predict(X_val))
knn_test_acc = accuracy_score(y_test, knn_model.predict(X_test))

y_test_pred_knn = knn_model.predict(X_test)

print(f"KNN Train Accuracy:      {knn_train_acc * 100:.2f}%")
print(f"KNN Validation Accuracy: {knn_val_acc * 100:.2f}%")
print(f"KNN Test Accuracy:       {knn_test_acc * 100:.2f}%")


# ============================================================
# 16. TRAIN AND EVALUATE: SUPPORT VECTOR MACHINE (SVM)
# ============================================================

print("\n" + "=" * 70)
print("TRAINING SVM CLASSIFIER")
print("=" * 70)

svm_model = SVC(kernel='rbf', C=10.0, gamma='scale', random_state=RANDOM_SEED)
svm_model.fit(X_train_combined, y_train_combined)

svm_train_acc = accuracy_score(y_train_combined, svm_model.predict(X_train_combined))
svm_val_acc = accuracy_score(y_val, svm_model.predict(X_val))
svm_test_acc = accuracy_score(y_test, svm_model.predict(X_test))

y_test_pred_svm = svm_model.predict(X_test)

print(f"SVM Train Accuracy:      {svm_train_acc * 100:.2f}%")
print(f"SVM Validation Accuracy: {svm_val_acc * 100:.2f}%")
print(f"SVM Test Accuracy:       {svm_test_acc * 100:.2f}%")


# ============================================================
# 17. FINAL COMPARISON & METRICS EVALUATION
# ============================================================

print("\n" + "=" * 70)
print("FINAL MODEL COMPARISON SUMMARY")
print("=" * 70)

print(f"{'Model':<10} | {'Test Acc':<10} | {'Macro F1':<10} | {'Weighted F1'}")
print("-" * 50)

knn_macro_f1 = f1_score(y_test, y_test_pred_knn, average="macro", zero_division=0)
knn_weighted_f1 = f1_score(y_test, y_test_pred_knn, average="weighted", zero_division=0)
svm_macro_f1 = f1_score(y_test, y_test_pred_svm, average="macro", zero_division=0)
svm_weighted_f1 = f1_score(y_test, y_test_pred_svm, average="weighted", zero_division=0)

print(f"{'KNN':<10} | {knn_test_acc*100:6.2f}%    | {knn_macro_f1:10.4f} | {knn_weighted_f1:10.4f}")
print(f"{'SVM':<10} | {svm_test_acc*100:6.2f}%    | {svm_macro_f1:10.4f} | {svm_weighted_f1:10.4f}")


# ============================================================
# 18. DETAILED REPORTS & CONFUSION MATRICES
# ============================================================

print("\n" + "=" * 70)
print("KNN CLASSIFICATION REPORT")
print("=" * 70)
print(classification_report(y_test, y_test_pred_knn, digits=4, zero_division=0))

print("\n" + "=" * 70)
print("SVM CLASSIFICATION REPORT")
print("=" * 70)
print(classification_report(y_test, y_test_pred_svm, digits=4, zero_division=0))

# Plot Confusion Matrices
fig, axes = plt.subplots(1, 2, figsize=(16, 7))

cm_knn = confusion_matrix(y_test, y_test_pred_knn, labels=np.arange(47))
axes[0].imshow(cm_knn, interpolation="nearest", cmap="Blues")
axes[0].set_title("KNN Confusion Matrix")
axes[0].set_xlabel("Predicted Label")
axes[0].set_ylabel("True Label")

cm_svm = confusion_matrix(y_test, y_test_pred_svm, labels=np.arange(47))
axes[1].imshow(cm_svm, interpolation="nearest", cmap="Greens")
axes[1].set_title("SVM Confusion Matrix")
axes[1].set_xlabel("Predicted Label")
axes[1].set_ylabel("True Label")

plt.tight_layout()
plt.show()

print("\nExperiment completed successfully.")
```
